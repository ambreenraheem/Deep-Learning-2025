Answer (b): **Data Augmentation : Modifying existing data to create new examples.**

- Examples like: Rotating, flipping, cropping, noise addition in images
- Source of data should be "Real data is altered"
- used in cases like: To prevent overfitting and increase dataset diversity

**Synthetic Data Approach: Creating entirely new data using simulations or generative models**

- Examples like: Using GANs to generate fake images, or simulated environments to create virtual data
- Source of data should be "No real data is needed, generated from scratch"
- used in cases like: When data is unavailable or expensive to collect

Answer (c): **Forward Propagation:
It is the process where input data passes through the network, layer by layer, to generate an output prediction. Each neuron applies a weight, bias, and activation function to the input.**

**Backward Propagation:
It is the process of updating weights and biases by computing gradients using the loss function. It involves the chain rule and helps the network learn by minimizing the error during training.**

Answer (d): **Machine Learning:**
- Works well with structured data
- Feature extraction is manual and it's Algorithms are Decision Trees, SVM, etc.
- Requires less computational power

**Deep Learning:**
- Works well with unstructured data (images, text)
- Learns features automatically using neural networks and Uses multi-layer neural networks (ANNs, CNNs, RNNs)
- Requires high computational power and data
